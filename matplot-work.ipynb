# sample matplot lib plotting

```python
%matplotlib inline
import matplotlib.pyplot as plt

# 1. prepare the data
x = [1,2,3,4]
y= [11,22,33,44]

# 2. setup plot
fig, ax = plt.subplots(figsize=(5,5))

# 3. plot data
ax.plot(x,y)

# 4. Customize plot 
ax.set(title='sd plot', xlabel='x-axis',ylabel='y-axis');
```
